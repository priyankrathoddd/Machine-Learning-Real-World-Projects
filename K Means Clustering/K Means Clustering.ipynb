```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Customer Data

customers = np.array([
    [20,500], # C1
    [40,1000], # C2
    [30,800], # C3
    [18,300], # C4
    [25,600], # C5
])


# Run KMean with 2 group

kmeans = KMeans(n_clusters=2,random_state=0) 
kmeans.fit(customers)
```

```python
# Get group centres

groups = kmeans.labels_
centers = kmeans.cluster_centers_
print("Number of Groups = ",groups)
print("Centers = ",centers)
```

```python
plt.scatter(customers[:,0],customers[:, 1],c=groups,s=50,cmap="viridis",label="Customers")
plt.scatter(centers[:,0],centers[:, 1],c="red",s=200,marker="X",label="centers")

plt.title("Our Shopping group")
plt.xlabel("Shops Per Month")
plt.ylabel("Spending (INR)")
plt.show()
```

```python
wcss = []

for k in range(1,5):
    kmeans = KMeans(n_clusters=k,random_state=0)
    kmeans.fit(customers)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,5),wcss,marker = "o")
plt.title("Elbow Trick")
plt.xlabel("Number Of Group")
plt.ylabel("Spending (INR)")
plt.show()
```
